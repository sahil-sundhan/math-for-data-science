```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv("AB_testing_data.csv")
```

```python
df.head()
```

```python
df.loc[:,['Conversion fraction A', 'Conversion fraction B']].describe()
```

Here mean of conversion fraction of B is more than that of A.
Shall we assume B is better than A?
</br>
No, we shall perform two sample t test to establish the statistical significance.

### Two sample t-test with equal varaiance

H0: Two samples are same, any difference is purely by chance.

```python
sample_1 = df['Conversion fraction A']
sample_2 = df['Conversion fraction B']
```

```python
eq_var = True
```

```python
stat, p = stats.ttest_ind(sample_1, sample_2, equal_var=eq_var)
print(f"t-test statistic: {stat}")
print(f"p-value: {p}")
print(f"{'H0 Rejected' if p < 0.05 else 'H0 Accepted'}")
```
