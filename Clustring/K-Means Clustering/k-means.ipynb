```python
numerical_features = ['no_of_adults', 'no_of_children', 'no_of_weekend_nights',
       'no_of_week_nights', 'required_car_parking_space', 'lead_time','repeated_guest',
       'avg_price_per_room', 'no_of_special_requests']
categorical_features  = ['type_of_meal_plan','room_type_reserved','market_segment_type']
```

```python
data_features = data[numerical_features + categorical_features + ['booking_status']]
```

```python
data_features.head()
```

### Step 3: Create features from the Data

```python
x_train = data_features[numerical_features + categorical_features]

# There is nothing to predict in clustering
# We are just storing booking status flag in another variable to 
# check later if the clusters have some pattern w.r.t booking cancellation

y_label = data_features[['booking_status']]
```

```python
x_train = pd.concat([x_train[numerical_features], pd.get_dummies(x_train[categorical_features]).replace({True: 1, False: 0})], axis=1).reset_index(drop=True)
```

```python
x_train.head()
```

### Step 4: Scale the data

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
```

```python
x_train[numerical_features] = scaler.fit_transform(x_train[numerical_features])
```

```python
x_train.to_csv('x_train.csv', index=False)
```

### Step 5: Apply K-Means Clustering

```python
from sklearn.cluster import KMeans
model = KMeans(n_clusters=10, random_state=0, n_init='auto')
model.fit(x_train)
```

```python
from sklearn.metrics import silhouette_score
silhouette_score(x_train, model.labels_)
```

```python
x_train_copy = x_train.copy()
```

```python
x_train['cluster_labels'] = model.labels_
x_train['booking_status'] = y_label['booking_status']
```

### Step 6: Understand patterns in clusters

```python
x_train.booking_status.value_counts()
```

```python
cancelled = x_train.booking_status.value_counts().iloc[1]
not_cancelled = x_train.booking_status.value_counts().iloc[0]
```

```python
print(f"Cancellation rate: {100*(cancelled/(cancelled + not_cancelled))}")
```

```python
cluster_number = list()
cancellation_rate = list()
```

```python
for cluster in x_train.cluster_labels.unique():
    temp_df = x_train.loc[x_train.cluster_labels == cluster]
    not_cncl = temp_df.booking_status.value_counts().iloc[0]
    cncl = temp_df.booking_status.value_counts().iloc[1]
    cncl_rate = (100*(cncl/(cncl + not_cncl)))
    cluster_number.append(cluster)
    cancellation_rate.append(cncl_rate)
```

```python
fig = plt.figure()
sns.barplot(data=pd.DataFrame(data={'cluster label': cluster_number, 'cancellation rate': cancellation_rate}), x='cluster label', y='cancellation rate')
plt.show()
```

```python
sns.countplot(data=x_train, x='booking_status', hue='cluster_labels')
```

### Step 7: Elbow curve

Ideally must be done before finalizing the clusters

```python
rss = list()

for i in range(2, 30):
    mdl = KMeans(n_clusters=i, init='k-means++', n_init='auto', random_state=0)
    mdl.fit(x_train_copy)
    rss.append(mdl.inertia_)
```

```python
rss_df = pd.DataFrame({'rss': rss, 'k': range(2, 30)})
```

```python
rss_df
```

```python
fig = plt.figure(figsize=(10, 4))
sns.pointplot(data=rss_df, x = 'k', y = 'rss', markers='x', linestyles='--')
plt.xlabel('Number of clusters')
plt.ylabel('Residual Sum of Squares')
plt.show()
```
