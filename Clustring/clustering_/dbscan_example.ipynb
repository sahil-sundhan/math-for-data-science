```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons, make_blobs
from sklearn.cluster import DBSCAN, KMeans
```

```python
# Generate two datasets: one with blobs, another with non-linear moon shapes
X1, _ = make_blobs(n_samples=300, centers=3, random_state=42, cluster_std=1.0)
X2, _ = make_moons(n_samples=300, noise=0.1, random_state=42)
```

```python
def plot_clusters(X, labels, title):
    plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=30)
    plt.title(title)
    plt.show()
```

```python
# K-Means Clustering
kmeans1 = KMeans(n_clusters=3, random_state=42).fit(X1)
kmeans2 = KMeans(n_clusters=2, random_state=42).fit(X2)
```

```python
# DBSCAN Clustering
dbscan1 = DBSCAN(eps=1.2, min_samples=5).fit(X1)
dbscan2 = DBSCAN(eps=0.2, min_samples=5).fit(X2)
```

```python
# Plot results
print("K-Means on Blob Data")
plot_clusters(X1, kmeans1.labels_, "K-Means: Blob Data")
```

```python
print("DBSCAN on Blob Data")
plot_clusters(X1, dbscan1.labels_, "DBSCAN: Blob Data")
```

```python
print("K-Means on Moon Data")
plot_clusters(X2, kmeans2.labels_, "K-Means: Moon Data")
```

```python
print("DBSCAN on Moon Data")
plot_clusters(X2, dbscan2.labels_, "DBSCAN: Moon Data")
```
